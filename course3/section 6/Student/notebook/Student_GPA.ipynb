<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
    در ابتدا نیاز است فایل‌های مجموعه‌داده را بخوانید. نمونه‌های آموزشی در فایل <code>train.csv</code> و نمونه‌های آزمون که باید مقدار متغیر هدف آن‌ها را پیش‌بینی کنید در فایل <code>test.csv</code> ذخیره شده‌اند. اگر لازم دانستید می‌توانید به دلخواه خود بخشی از مجموعه‌ی آموزش را به عنوان مجموعه‌ی اعتبارسنجی نیز جدا کنید.

```python
import pandas as pd
import numpy as np
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn import linear_model
from sklearn import tree
from sklearn import ensemble
from sklearn import svm
from sklearn import neighbors
from sklearn import metrics
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
```

```python
# Reading/Loading the dataset files
data_train = pd.read_csv("../data/train.csv")
data_test = pd.read_csv("../data/test.csv")
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
پیش‌پردازش و مهندسی ویژگی
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
        در این سوال شما می‌توانید از هر تکنیک پیش‌پردازش/مهندسی ویژگی دلخواهتان، استفاده کنید.
    <br>
    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت؛ بنابراین هر چه پیش‌پردازش/مهندسی ویژگی بهتری انجام دهید تا دقت مدل بهبود پیدا کند، امتیاز بیشتری از این سوال کسب خواهید کرد.

</font>
</p>

```python
# Preprocessing step
data_train = data_train.drop("StudentID",axis=1)
data_test = data_test.drop("StudentID",axis=1)

data_train.info()
```

```python
data_train.describe().T
```

```python
data_train["Ethnicity"].value_counts()
```

```python
print(data_train["ParentalEducation"].value_counts())
print("------------------------------------------")

print(data_train["ParentalSupport"].value_counts())
print("------------------------------------------")

mode_ParentalEducation = data_train["ParentalEducation"].mode()[0]
print(mode_ParentalEducation)

print("------------------------------------------")
mode_ParentalSupport = data_train["ParentalSupport"].mode()[0]
print(mode_ParentalSupport)
```

```python
print(data_train.isnull().sum())
```

```python
data_train["ParentalEducation"] = data_train["ParentalEducation"].fillna(mode_ParentalEducation)

data_train["ParentalSupport"] = data_train["ParentalSupport"].fillna(mode_ParentalSupport)

data_test["ParentalEducation"] = data_test["ParentalEducation"].fillna(mode_ParentalEducation)

data_test["ParentalSupport"] = data_test["ParentalSupport"].fillna(mode_ParentalSupport)
```

```python
data_train['ParentalSupport']=data_train['ParentalSupport'].replace({
    "Low":0, "Moderate":1, "High":2, "Very High":3}).astype(int)

data_train['ParentalEducation']=data_train['ParentalEducation'].replace({
    "High School":0, "Some College":1, "Bachelor":2, "Higher":3}).astype(int)

data_test['ParentalSupport']=data_test['ParentalSupport'].replace({
    "Low":0, "Moderate":1, "High":2, "Very High":3}).astype(int)

data_test['ParentalEducation']=data_test['ParentalEducation'].replace({
    "High School":0, "Some College":1, "Bachelor":2, "Higher":3}).astype(int)
```

```python
le=preprocessing.LabelEncoder()

data_train["Ethnicity"]=le.fit_transform(data_train["Ethnicity"])

data_test["Ethnicity"]=le.transform(data_test["Ethnicity"])
```

```python
data_train.corr()["GPA"]
```

```python
x = data_train.drop("GPA",axis=1)
y = data_train["GPA"]
```

```python
x_train, x_valid, y_train, y_valid = train_test_split(
    x, y, test_size=0.2, random_state=42
)

x_train.shape, x_valid.shape, y_train.shape, y_valid.shape
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
آموزش مدل
</font>
</h2>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazir" size=3>
    حال که داده را پاکسازی کرده و احتمالا ویژگی‌هایی را به آن افزوده یا از آن حذف کرده‌اید، وقت آن است که مدلی آموزش دهید که بتواند متغیر هدف این مسئله را پیش‌بینی کند.
</font>
</p>

```python
# Model design
model_1 = linear_model.LinearRegression()
model_1.fit(x_train,y_train)

model_2 = tree.DecisionTreeRegressor()
model_2.fit(x_train,y_train)

model_3 = ensemble.RandomForestRegressor()
model_3.fit(x_train,y_train)

model_4 = svm.SVR()
model_4.fit(x_train,y_train)

model_5 = neighbors.KNeighborsRegressor(20)
model_5.fit(x_train,y_train)

model_6 = Sequential()
model_6.add(Dense(32,activation="relu",input_shape=x_train.iloc[0].shape))
model_6.add(Dense(1,activation="linear"))
model_6.compile(optimizer="adam", loss="mae")
model_6.fit(x_train, y_train, epochs=20, validation_data=(x_valid,y_valid))
model_6.summary()
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
معیار ارزیابی
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
    معیاری که برای ارزیابی عملکرد مدل انتخاب کرده‌ایم، <code>r2_score</code> نام دارد.
    <br>
    این معیار، سنجه ارزیابی کیفیت مدل شماست. به عبارت بهتر در سامانه داوری هم از همین معیار برای نمره‌دهی استفاده شده است.
    <br>
    پیشنهاد می‌شود با توجه به این معیار، عملکرد مدل خود را بر روی مجموعه‌ی آموزش یا اعتبارسنجی ارزیابی کنید.
</font>
</p>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font color="red"><b color='red'>توجه:</b></font>
<font face="vazir" size=3>
    برای دریافت نمره از این سوال لازم است تا دقت مدل شما از آستانه‌ی ۰.۴ بیشتر باشد.
    در صورتی که دقت مدل شما از ۰.۴ کمتر باشد نمره شما 
    <b>صفر</b>
    خواهد شد و در غیر این صورت با فرمول زیر محاسبه می‌شود:
</font>
</p>


```python
# evaluate your model
from sklearn.metrics import r2_score

# داده را به آموزش و ارزیابی تقسیم کن
# مدل را روی داده های validation ارزیابی کنید
pred_1 = model_1.predict(x_valid)
res_1 = metrics.r2_score(y_valid,pred_1)
pred_2 = model_2.predict(x_valid)
res_2 = metrics.r2_score(y_valid,pred_2)
pred_3 = model_3.predict(x_valid)
res_3 = metrics.r2_score(y_valid,pred_3)
pred_4 = model_4.predict(x_valid)
res_4 = metrics.r2_score(y_valid,pred_4)
pred_5 = model_5.predict(x_valid)
res_5 = metrics.r2_score(y_valid,pred_5)
pred_6 = model_6.predict(x_valid)
res_6 = metrics.r2_score(y_valid,pred_6)

res_list = [res_1,res_2,res_3,res_4,res_5,res_6]
for i in range(6):
    print(f"r2-score for model_{i+1} = ",res_list[i])
```

```python
import zipfile
import os

if not os.path.exists(os.path.join(os.getcwd(), 'Student_GPA.ipynb')):
    %notebook -e Student_GPA.ipynb

def compress(file_names):
    print("File Paths:")
    print(file_names)
    compression = zipfile.ZIP_DEFLATED
    with zipfile.ZipFile("result.zip", mode="w") as zf:
        for file_name in file_names:
            zf.write('./' + file_name, file_name, compress_type=compression)

submission.to_csv('submission.csv', index=False)

file_names = ['Student_GPA.ipynb', 'submission.csv']
compress(file_names)
```
