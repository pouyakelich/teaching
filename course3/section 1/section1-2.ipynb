```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
data = pd.read_csv("easy_lung_cancer.csv")
data
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
data.duplicated().sum()
```

```python
data["gender"] = data["gender"].replace({"M": 0, "F": 1})
data["lung cancer"] = data["lung cancer"].replace({"NO": 0, "YES": 1})
data = data.astype('float32')
```

```python
x = data.drop("lung cancer",axis=1)
y = data["lung cancer"]
```

```python
data.shape
```

```python
x.iloc[0].shape
```

```python
model = Sequential()
model.add(Dense(units=32 , activation="relu" , input_shape=(6,)))
model.add(Dense(units=1 , activation="sigmoid"))
model.summary()
```

```python
model.compile(optimizer="adam" , 
              loss="binary_crossentropy" ,
              metrics = ["accuracy"])
```

```python
model.fit(x,y,epochs=10)
```
