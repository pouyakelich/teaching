Hidden Layer in Neural Networks                          
A Hidden Layer in an artificial neural network (ANN) is a layer of neurons located between the Input Layer and the Output Layer. It is termed "hidden" because its inputs and outputs are internal to the network and are not directly observable from the environment.                                    
1. Primary FunctionThe main purpose of a hidden layer is feature extraction and learning complex, non-linear relationships within the data:                        
Linear Problems: Can be solved without hidden layers by mapping inputs directly to outputs using a straight decision boundary.                                                
Non-Linear Problems (Real-world Data): Require hidden layers combined with non-linear activation functions (such as ReLU or Sigmoid) to allow the network to learn intricate patterns and high-level representations.                               
2. How it Works (Step-by-Step)                            
Within each neuron of a hidden layer, the following operations occur:                                        
1. Linear Combination: Inputs from the previous layer ($x$) are multiplied by their respective weights ($w$) and added to a bias term ($b$):                                    
$$z = \sum_{i=1}^{n} w_i x_i + b = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b$$
2. Activation Function: The weighted sum $z$ is passed through a non-linear activation function $f(z)$ (e.g., ReLU):$$a = f(z) = \max(0, z)$$Forward Propagation:                  
 The resulting activation value $a$ is passed forward as an input to the subsequent layer.

4. Key Considerations                      
Network Depth: Networks with $1$ or $2$ hidden layers are considered shallow neural networks, whereas networks with multiple hidden layers form the basis of Deep Learning.        
Model Capacity & Overfitting:                        
Too few neurons/layers: May lead to underfitting (the model fails to capture complex patterns).                  
Too many neurons/layers: May lead to overfitting (the model memorizes noise in the training data and generalizes poorly to new data).

The Sigmoid Function                          
The Sigmoid Function is a widely used activation function in artificial neural networks that squashes any real-valued number into a range between 0 and 1.                                          
1. Mathematical Formula$$\sigma(x) = \frac{1}{1 + e^{-x}}$$       
2. Key Purposes and Applications                      
. Probability Interpretation: Because the output is strictly bounded between 0 and 1, it can be interpreted directly as a probability (e.g., an output of $0.85$ indicates an $85\%$ chance of a positive classification).                             
. Binary Classification Output Layer: It is predominantly used in the final output layer for binary classification problems (such as predicting whether a patient has lung cancer or not) to yield a probabilistic decision.                               
. S-shaped Curve: The curve smoothly maps extreme negative inputs toward $0$ and extreme positive inputs toward $1$, providing a smooth gradient for optimization.

```python
model = Sequential()
#making our hidden layer
model.add(Dense(units=32 , activation="relu" , input_shape=(6,)))
#output layer
model.add(Dense(units=1 , activation="sigmoid"))
model.summary()
```

1. Optimizer (optimizer="adam")                         
Role: The optimizer determines how the weights and biases of the network are updated during backpropagation to minimize the loss.                                      
Adam (Adaptive Moment Estimation): One of the most popular and efficient optimization algorithms in deep learning. It automatically adapts the learning rates for each parameter, resulting in fast convergence and reliable training performance.                                   
2. Loss Function (loss="binary_crossentropy")                                   
Role: The loss function measures the penalty or error between the model's predicted output and the true target value.                                   
Binary Crossentropy: Specifically designed for binary classification tasks (e.g., predicting whether a patient has lung cancer ($1$) or not ($0$)). It evaluates how well the predicted probability (output of the sigmoid function) matches the actual binary label.                                
3. Metrics (metrics=["accuracy"])                                      
Role: Metrics are used to monitor and evaluate the performance of the model during training and testing. Unlike the loss function, they are not used directly to update network weights.                              
Accuracy: Represents the percentage of correctly classified samples out of the total samples evaluated (e.g., an accuracy of $0.85$ means $85\%$ of the predictions were correct).

```python
model.compile(optimizer="adam" , 
              loss="binary_crossentropy" ,
              metrics = ["accuracy"])
```

```python
model.fit(x,y,epochs=10)
```
