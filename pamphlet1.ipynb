Now we wanna learn about OR,AND

OR:

🟢True or True = True                     
🟢False or True = True                      
🟢True or False = True                        
🔴False or False = False

```python
Home = True
Car = True

if Home or Car:
    print("Nice 1")

Home = True
Car = False

if Home or Car:
    print("Nice 2")


Home = False
Car = True

if Home or Car:
    print("Nice 3")

Home = False
Car = False

if Home or Car:
    print("Nice 4")
```

AND:

🟢True or True = True                     
🔴False or True = False                      
🔴True or False = False                        
🔴False or False = False

```python
Home = True
Car = True

if Home and Car:
    print("Nice 1")

Home = True
Car = False

if Home and Car:
    print("Nice 2")


Home = False
Car = True

if Home and Car:
    print("Nice 3")

Home = False
Car = False

if Home and Car:
    print("Nice 4")
```

Now we wanna make a simpe calculator

```python
num1 = float(input())
num2 = float(input())
operator = input("+ / - *")

if operator == "+":
    print(num1+num2)
elif operator == "-":
    print(num1-num2)
elif operator == "*":
    print(num1*num2)
else:
    print(num1/num2)
```
