Water exists in 3 different states based on its temperature:

🟣Answer to Assignment 3:

We use an infinite while loop and break when the user enters 0. We keep a counter and a sum.

```python
count = 0
total = 0

while True:
    number = float(input("Enter a number (0 to stop): "))

    if number == 0:
        break

    count += 1
    total += number

print(f"You entered {count} numbers and their sum is {total}.")
```

❗Tip: The counter and the sum must be created before the loop, not inside it.

🟣Answer to Assignment 4:

We keep the items in a dictionary. The key is the name of the item and the value is its price.

```python
import random

shopping = {}

while True:
    command = input("add/remove/end: ")

    if command == "add":
        item = input("Enter the item name: ")
        price = float(input("Enter the price: "))
        shopping[item] = price

    elif command == "remove":
        item = input("Enter the item name: ")
        if item in shopping:
            del shopping[item]
        else:
            print("This item is not in the list.")

    else:
        break

print(shopping)
print("Number of items:", len(shopping))
print("Total price:", sum(shopping.values()))

# choosing today's offer randomly
if len(shopping) > 0:
    offer = random.choice(list(shopping.keys()))
    print("Today's offer:", offer, shopping[offer])
```

❗Tip: random.choice() needs a sequence, so we change the keys of the dictionary into a list first.

🟣Answer to Assignment 5:

Each function does one job: one calculates the average, and the other tells the status.

```python
def average(scores):
    return sum(scores) / len(scores)


def status(score):
    if score >= 12:
        return "passed"
    else:
        return "failed"


scores = []

for i in range(5):
    s = float(input("Enter a score: "))
    scores.append(s)

avg = average(scores)
print("Average:", avg)
print("Status:", status(avg))
```

❗Tip: We can also write the average without sum() by using a loop:

```python
def average(scores):
    total = 0
    for s in scores:
        total += s
    return total / len(scores)


average([15, 12, 18, 9, 20])
```

The class keeps the owner and the balance, and the methods work on the balance of that exact object.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print(f"{self.owner}: not enough money!")

    def show(self):
        print(f"{self.owner} -> {self.balance}")


a1 = BankAccount("Ali", 1000)
a2 = BankAccount("Sara", 2500)
a3 = BankAccount("Reza", 500)

a1.deposit(500)
a2.withdraw(1000)
a3.withdraw(800)      # not enough money

a1.show()
a2.show()
a3.show()
```

Now we sort the accounts by their balance with a lambda.

```python
accounts = [a1, a2, a3]

accounts = sorted(accounts, key=lambda acc: acc.balance, reverse=True)

for acc in accounts:
    acc.show()
```

❗Tip: sorted() returns a new list, so we save the result in a variable.
