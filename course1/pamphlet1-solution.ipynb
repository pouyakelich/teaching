Assignment1:                                                      
Create a program that asks the user to enter their name and their age. Print out a message addressed to them that tells them the year that they will turn 100 years old.                               
for example:                                               
pouya, you will be 100 years old in the year 2102

```python
name = input("What is your name: ")
age = int(input("How old are you: "))
year = 2014 - age + 100
print(name + ", you will be 100 years old in the year " + str(year))
```

Assignment 2:
Water exists in 3 different states based on its temperature:

    If it is above 100 degrees Celsius, it is steam.

    If it is between 0 and 100 degrees Celsius, it is water.

    If it is below 0 degrees Celsius, it is ice.

    Write a program that gets the temperature and tells us the water type

```python
temperature = float(input("Enter the temperature in Celsius: "))

if temperature > 100:
    print("Steam")
elif 0 <= temperature <= 100:
    print("Water")
else:
    print("Ice")
```
