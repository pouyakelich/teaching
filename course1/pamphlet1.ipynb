Example: A simple login system with a maximum of 3 attempts using a while loop and a break statement.

Tip: We use chances -= 1 to decrement the remaining attempts each time the user enters an incorrect password.

```python
correct_password = "Pouya Kelich"
chances = 3

while chances > 0:
    entered_password = input("Enter the password: ")
    if entered_password == correct_password:
        print("Access granted!")
        break
    else:
        chances -= 1
        if chances > 0:
            print(f"Incorrect password. You have {chances} chance(s) left.")
        else:
            print("Access denied. No chances left.")
```

Introduction to the random Module

Python has a built-in module called random that allows you to generate pseudo-random numbers, make random selections, and shuffle sequences.                     
To use it, you must first import the module:

```python
import random
```

Key Functions with Examples

1. random.randint(a, b)                          
Generates a random integer between a and b (inclusive — both endpoints can be chosen).

```python
num = random.randint(1, 10)
num
```

2. random.random()                                    
Returns a random floating-point number in the range [0.0, 1.0).

```python
val = random.random()
val
```

3. random.choice(sequence)                           
Selects and returns a random element from a non-empty sequence (like a list or tuple).

```python
fruits = ["apple", "banana", "cherry"]
picked = random.choice(fruits)
picked
```

Number Guessing Game:                 
 In this simple game, the computer picks a random number within a specified range, and the player tries to guess it. The program gives feedback ("Too high" or "Too low") after each guess until the player finds the correct number.

```python
import random

# Generate a random number between 1 and 100
secret_number = random.randint(1, 100)
attempts = 0

print("Welcome to the Number Guessing Game!")
print("I'm thinking of a number between 1 and 100.")

while True:
    guess = int(input("Enter your guess: "))
    attempts += 1

    if guess < secret_number:
        print("Too low! Try again.")
    elif guess > secret_number:
        print("Too high! Try again.")
    else:
        print(f"Congratulations! You guessed it in {attempts} attempts.")
        break
```

Definition of Tuples                                  
A tuple is a built-in data structure in Python used to store a collection of items in a single variable. Tuples are defined by placing all the elements inside parentheses (), separated by commas.

Key Characteristics

        Ordered: Items in a tuple maintain a fixed sequence, and their order cannot be changed.

        Unchangeable (Immutable): Once a tuple is created, you cannot add, remove, or modify its elements.

        Allows Duplicates: Like lists, tuples can contain duplicate values.

        Heterogeneous: A tuple can store items of different data types (e.g., integers, strings, floats, booleans).

```python
t = (3,5,7,9)
```

```python
t[0]
```

```python
t[2] = 3
```

Definition of Sets                            
A set is a built-in data structure in Python used to store an unordered collection of unique elements. Sets are defined by placing items inside curly braces {} separated by commas, or by using the set() function.

Key Characteristics

        Unordered: Items in a set do not have a defined order, and their sequence can change each time you use them.

        Unindexed: You cannot access elements by referring to an index or key (e.g., my_set[0] is invalid).

        Unique Elements Only: Sets automatically remove any duplicate values.

        Mutable Collection, Immutable Items: You can add or remove items from a set, but the individual elements themselves must be of an immutable data type (like numbers, strings, or tuples).

```python
s = {3 , 4 , 5 , 3 , 4 , 12 }
```

```python
s
```

```python
s[0]
```

Definition of Dictionaries                               

    A dictionary is a built-in data structure in Python used to store data values in key:value pairs. Dictionaries are written with curly brackets {} and contain keys and values separated by colons :.

Key Characteristics

        Key-Value Pairs: Every element consists of a unique key mapped to a specific value.

        Changeable (Mutable): You can add, modify, or remove key-value pairs after the dictionary is created.

        Unique Keys: Keys cannot be duplicated. If a duplicate key is provided during creation or assignment, the old value will be overwritten.

        Heterogeneous: Keys and values can be of any data type (though keys must be of an immutable type, like strings, integers, or tuples).

        Ordered: As of Python 3.7, dictionaries maintain the insertion order of items.

```python
d = {
    'math': 20,
    'Science': 14,
    'Sports' : 18
}
```
