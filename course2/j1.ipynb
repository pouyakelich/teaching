```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
list_ghad = [145,150,155,160,165,170,175,180,185,190]
list_vazn = [42, 45, 50, 55, 60, 65, 70, 76, 82, 87]
```

# Matplotlib

```python
plt.scatter(list_ghad,list_vazn)
plt.xlabel("ghad")
plt.ylabel("vazn")
plt.title("ghad va vazn")
plt.grid()
plt.show()
```

```python
plt.plot(list_ghad,list_vazn,color="#B82BA7")
plt.xlabel("ghad")
plt.ylabel("vazn")
plt.title("ghad va vazn")
plt.grid()
plt.show()
```

```python
plt.bar(list_ghad,list_vazn,color="#B82BA7",width=0.9)
plt.xlabel("ghad")
plt.ylabel("vazn")
plt.title("ghad va vazn")
plt.grid()
plt.show()
```

```python
plt.barh(list_ghad,list_vazn,color="#B82BA7",height=0.8)
plt.xlabel("vazn")
plt.ylabel("ghad")
plt.title("ghad va vazn")
plt.grid()
plt.show()
```

```python
plt.hist(list_ghad,bins=4)
plt.xlabel("ghad")
plt.ylabel("tedad")
plt.show()
```

```python
list_mobile = ["iphone","samsung","xiaomi",
               "nokia","huawei","other"]
list_percent = [32, 19, 10, 2, 1, 36]
plt.pie(list_percent,labels=list_mobile)
plt.legend(loc="right")
plt.show()
```
