```python
import numpy
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Now we wanna create our dataset

```python
height_list = [145,150,155,160,165,170,175,180,185,190]
weight_list = [42, 45, 50, 55, 60, 65, 70, 76, 82, 87]
```

# Matplotlib

This is a scatter chart

```python
plt.scatter(height_list,weight_list)
plt.xlabel("height")
plt.ylabel("weight")
plt.title("height and weight")
plt.grid()
plt.show()
```

This is a plot chart

```python
plt.plot(height_list,weight_list)
plt.xlabel("height")
plt.ylabel("weight")
plt.title("height and weight")
plt.grid()
plt.show()
```

This is a bar chart

```python
plt.bar(height_list,weight_list,color="#B82BA7",width=0.9)
plt.xlabel("height")
plt.ylabel("weight")
plt.title("height and weight")
plt.grid()
plt.show()
```

This is a horizontal bar chart

```python
plt.barh(height_list,weight_list,height=0.9)
plt.xlabel("height")
plt.ylabel("weight")
plt.title("height and weight")
plt.grid()
plt.show()
```

This is histogram chart

```python
plt.hist(height_list,bins=4)
plt.xlabel("height")
plt.ylabel("weight")
plt.show()
```

This is circle chart

```python
mobile_list = ["iphone","samsung","xiaomi",
               "nokia","huawei","other"]
percentage_list = [32, 19, 10, 2, 1, 36]
plt.pie(percentage_list,labels=mobile_list)
plt.legend(loc="right")
plt.show()
```

# Pandas

Lets make a dictionary then change it to pandas dataframe

```python
height_and_weight = {
    "height":height_list,
    "weight":weight_list
}
```

```python
data = pd.DataFrame(height_and_weight)
data
```

now we wanna add new column to our dataframe

```python
data["Name"] = ["Emma","Rose","john",
               "leonard","jack","richard",
               "william","angela","rock",
               "maria"]
data
```

If we dont know the value of something we can use NaN or None or Null

```python
data["age"] = [10,15,14,15,None,13,None,16,20,None]
data
```

Now we can save our dataframe into csv file

Index value is for making number for each row

```python
data.to_csv("height and weight.csv",index=False)
```
