```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Now we have a csv file and we want to read it

```python
data = pd.read_csv("Salary Data.csv")
data
```

Now make a scatter chart for this dataframe

```python
plt.scatter(data["Experience Years"] , data["Salary"])
plt.title("experience vs salary")
plt.xlabel("experience")
plt.ylabel("salary")
plt.grid()
plt.show()
```

Now make a plot chart for this dataframe

```python
plt.plot(data["Experience Years"],data["Salary"])
plt.title("experience vs salary")
plt.xlabel("experience")
plt.ylabel("salary")
plt.grid()
plt.show()
```

Now make a bar chart for this dataframe

```python
plt.bar(data["Experience Years"],data["Salary"])
plt.title("experience vs salary")
plt.xlabel("experience")
plt.ylabel("salary")
plt.grid()
plt.show()
```
