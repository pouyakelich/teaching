```python
y = df["FinalScore"]
y
```

```python
model = linear_model.LinearRegression()
```

```python
model.fit(x,y)
```

```python
features = ["study hour", "Attendance", "PreviousGrade"]
input_list = []
for i in range(3):
    your_input = float(input(f"Enter {features}: "))
    input_list.append(your_input)
score = model.predict([input_list])

print("\n--- Results ---")
print("study hour =", input_list[0])
print("Attendance =", input_list[1])
print("PreviousGrade =", input_list[2])
print("predicted final score =", score[0])
```
