Now we want to train our first modle;)

```python
x = data[["Experience Years"]]
y = data["Salary"]
```

We want to use linear regression modle

```python
model = linear_model.LinearRegression()
```

In this step modle learn the pattern between x and y(input and output)

```python
model.fit(x,y)
```

Now we wanna see our salary based on our Work Experience

```python
Work_Experience = float(input("how many years do you work?"))
print("your work experience:", Work_Experience)

salary = model.predict([[Work_Experience]])

print("your predicted salary:", float(salary[0]))
```

Why use double square brackets [[Work_Experience]]?

In scikit-learn, the .predict() method expects data in a 2D array shape (n_samples, n_features):

Outer brackets [ ] represent the list of samples (data rows).

Inner brackets [ ] represent the features (data columns) for a specific sample.

➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖

Why use salary[0] instead of salary?

In scikit-learn, the .predict() method always returns a NumPy array of predictions, even if you pass a single input sample.

salary represents the entire output array: array([50000.0])

salary[0] extracts the actual numerical value stored at index 0: 50000.0

Using float(salary[0]) ensures you extract the scalar float value cleanly without array formatting.

# Mean Absolute Error (MAE)


MAE measures the average magnitude of errors between actual values (y) and predicted values (y_predict), ignoring error direction (positive vs. negative).

y: Actual ground truth values.

y_predict: Predicted values generated by the model.
